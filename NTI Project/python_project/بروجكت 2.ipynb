```python
for col in dataset.select_dtypes(include=['string']).columns:
    dataset[col] = dataset[col].astype(str).str.replace(' ', '', regex=False)
```

```python
dataset.duplicated().sum() 
```

```python
dataset['Delivery_person_Age'].value_counts()
```

```python
dataset['Delivery_person_Age'] = dataset['Delivery_person_Age'].replace('NaN', '28')
dataset['Delivery_person_Age'] = dataset['Delivery_person_Age'].astype('Int64')
```

```python
dataset['Delivery_person_Age'].value_counts()
```

```python
dataset['Delivery_person_Ratings'].unique()
```

```python
dataset['Delivery_person_Ratings'] = dataset['Delivery_person_Ratings'].str.replace('NaN', '4.6')
dataset['Delivery_person_Ratings'] = dataset['Delivery_person_Ratings'].astype('float64')
```

```python
dataset['Delivery_person_Ratings'].unique()
```

```python
dataset[['Restaurant_latitude', 'Restaurant_longitude']].describe()
```

```python
dataset['Restaurant_latitude'] = dataset['Restaurant_latitude'].abs()
dataset['Restaurant_longitude'] = dataset['Restaurant_longitude'].abs()
```

```python
(dataset['Restaurant_latitude'] < 1).sum()
```

```python
(dataset['Restaurant_longitude'] < 1).sum()
```

```python
dataset.loc[dataset['Restaurant_latitude'] < 1,
            ['Restaurant_latitude', 'Restaurant_longitude']] = None
```

```python
dataset[['Restaurant_latitude', 'Restaurant_longitude']].isnull().sum()
```

```python
dataset[['Delivery_location_latitude', 'Delivery_location_longitude']].describe()
```

```python
dataset['Order_Date'] = pd.to_datetime(dataset['Order_Date'], dayfirst=True, format='mixed')
```
